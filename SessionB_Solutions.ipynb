```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
```

## Loading the Dataset

We load the penguins dataset from seaborn.

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

## Histograms — Identifying Distribution Shape

A histogram is the simplest tool for visualising the distribution of a variable.

Below we create a histogram of `bill_length_mm`.

```python
plt.figure(figsize=(7,4))
sns.histplot(data=penguins, x="bill_length_mm", bins=20, color="steelblue")
plt.show()
```

### Exercise 1  
Looking at the histogram above, describe the distribution of bill length.  
Is it symmetric, skewed, or multimodal?

Your Answer:

The distribution of penguin bill length is not perfectly symmetric or bell-shaped and appears approximately bimodal, with two broad concentrations of values.Most observations fall between approximately 35 and 52 mm, with two broad concentrations around 38–41 mm and 45–51 mm. This suggests a potentially bimodal distribution, although the peaks are not clearly separated.  The dataset may contain groups with different typical bill lengths, which becomes clearer when the species are examined separately.

## 2. Density Plots — Smooth Distribution Shapes

A density plot gives a smooth approximation of a histogram.

```python
plt.figure(figsize=(7,4))
sns.kdeplot(data=penguins, x="bill_length_mm", fill=True, alpha=0.4, color="orange")
plt.show()
```

### Exercise 2  
Compare the histogram and the density plot.  
What additional detail does the density plot reveal, if any?

Your Answer:

The histogram displays counts within bins, so it makes the number of observations in different bill-length intervals clear. The density plot is a smoothed estimate of the same distribution. It makes the broad peaks and the dip between them easier to see, although smoothing can hide some detail and its appearance depends on the bandwidth. 

## 3. Comparing Groups Using Faceted Histograms

We often want to compare the distribution of a variable across groups.

Here you will create separate histograms of bill length for each penguin species.

```python
g = sns.FacetGrid(penguins, col="species", height=4, sharex=True, sharey=True)
g.map_dataframe(sns.histplot, x="bill_length_mm", bins=20, color="steelblue")
plt.show()
```

### Exercise 3  
Describe one difference between the species' bill length distributions.

Your Answer:

Adelie penguins generally have shorter bills than Chinstrap and Gentoo penguins. Their histogram is concentrated at lower bill lengths. Chinstrap and Gentoo tend to have longer bills, although their distributions overlap. Separating the species also helps explain why the combined histogram has multiple concentrations.

## 4. Scatterplots — Visualising Relationships

## 5. Normality Checking Using Q–Q Plots

Python does not include Q–Q plots in seaborn, so we use scipy.stats.

```python
import scipy.stats as stats
```

A Q–Q plot helps determine whether a variable follows a normal distribution.
If the points lie close to the diagonal line, the distribution is approximately normal.

```python
plt.figure(figsize=(6,6))
stats.probplot(penguins["bill_length_mm"].dropna(), dist="norm", plot=plt)
plt.title("Q-Q Plot: Bill Length")
plt.show()
```

### Exercise 5  
Does bill length appear normally distributed?  
Explain how you know using the Q–Q plot.

Your Answer:

The Q–Q plot shows that many observations follow the theoretical normal distribution reasonably closely, particularly across portions of the centre. However, there are systematic departures from the reference line, including curvature through the lower, middle and upper quantiles. This suggests that the distribution of bill length is not perfectly normal. Differences between penguin species may contribute to these deviations. A Q–Q plot is a visual diagnostic, not a formal hypothesis test.
